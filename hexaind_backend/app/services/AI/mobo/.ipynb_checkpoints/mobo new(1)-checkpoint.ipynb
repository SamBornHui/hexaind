```python
from pydantic import BaseModel, Field, constr
from typing import List, Optional, Union, Dict
from enum import Enum

class TabularColumnType(str, Enum):

    NUMERICAL = 'NUMERICAL'

    CATEGORICAL = 'CATEGORICAL'
 
class ColumnStatistics(BaseModel):
    name: str
    type: TabularColumnType
    parameters: dict
 
class Visualization(BaseModel):
    x_axis: str
    y_axis: str
    graph: str
    graph_type: str
    color_by: str
 
class DatasetSchema(BaseModel):
    column_name: str
    data_type: str
 

class TabularDatasetInformation(BaseModel):
    preview: Union[dict, None]  # Define this more precisely based on the structure of 'preview'
    statistics: Union[List[ColumnStatistics], None] = Field(default=None)
    visualize: Union[List[Visualization], None] = Field(default=None)
    schema: Union[List[DatasetSchema], None] = Field(default=None)
```

```python
import pandas as pd

df = pd.read_csv('tests/bced_v2_initial.csv')
df
```

```python
def get_column_statistics_from_data_frame(input_data: pd.DataFrame) -> List[ColumnStatistics]:
        
        if not isinstance(input_data, pd.DataFrame):
            raise Exception("Expecting the input_data as pandas dataframe")
        
        column_statistics = []

        description = input_data.describe(include=['int16', 'int32', 'int64', 'float16', 'float32', 'float64', 'object', 'category']) # Extend the types if needed (don't use all, since some data types may generate exceptions)

        for column in input_data.columns:
            column_type = TabularColumnType.NUMERICAL if input_data[column].dtype in ['int16', 'int32', 'int64', 'float16', 'float32', 'float64'] else TabularColumnType.CATEGORICAL
            stats = description[column].to_dict()

            column_stat = ColumnStatistics(name=column, type=column_type, parameters=stats)
            column_statistics.append(column_stat)
        
        return column_statistics
```

```python
from pydantic import BaseModel, Field

from enum import Enum

from typing import List, Optional

class MOBOExperimentType(str, Enum):

    OPTIMIZATION = "OPTIMIZATION"

    ACTIVE_LEARNING = "ACTIVE_LEARNING"

class MOBOExperimentOutputVariablesObjectives(str, Enum):

    MINIMUM = "MINIMUM"

    MAXIMUM = "MAXIMUM"

class MOBOConfig(BaseModel):

    version: Optional[str] = Field(default="1.0", description="Version of the workflow")

    experiment_type: MOBOExperimentType = Field(default=MOBOExperimentType.OPTIMIZATION, description="mobo algorithm type")

    input_variables: List[str] = Field(..., description="input features for mobo")

    input_variables_constraints: List[List[float]] = Field(..., description="filters on input features")

    output_variables: List[str] = Field(..., description="output features for mobo")

    output_variables_objectives: List[MOBOExperimentOutputVariablesObjectives] = Field(..., description="user defined mobo experiment name ")

    output_variables_thresholds: List[float] = Field(..., description="thresholds for the output features, this is more like boundary for recommendations")

    num_iterations: int = Field(..., description="number of times to run the experiment or target no.of batches")

    batch_size: int = Field(..., description="specifies no.of recommendations will be generated in one batch")

    constraints_module_id: Optional[str] = Field(default=None, description="module id for the user uploaded mobo recommendations validation scripts")

    outcome_constraints_active: Optional[bool] = Field(default=False, description="Outcome constraints tog ")

    outcome_constraints_variables: Optional[List[str]] = Field(default=[], description="user defined mobo experiment name ")

    outcome_constraints_operator: Optional[str] = Field(default=None, description="user defined mobo experiment name ")

    outcome_constaints_inputvalue: Optional[int] = Field(default=None, description="user defined mobo experiment name ")

```

```python
ad = MOBOConfig(**mobo_input_payload)
ad
```

```python
cc = ad.model_dump()
```

```python
cc['experiment_type'] == 'OPTIMIZATION'

objective_min_bool = []
for i in cc['output_variables_objectives']:
    if i == 'MINIMUM':
        objective_min_bool.append(True)
    elif i == 'MAXIMUM':
        objective_min_bool.append(False)

    print(i)

objective_min_bool
```
