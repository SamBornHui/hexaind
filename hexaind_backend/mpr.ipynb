```python
import pandas as pd 
def convert_predictions_to_df(predictions, output_cols, output_std_cols):
        # Check if predictions contain dictionaries or lists
        
        # if isinstance(predictions, list):
        #     df_preds = pd.DataFrame(preds_data, columns=output_cols)
        #     print(df_preds)
        
        if isinstance(predictions[0], dict):
            df = pd.DataFrame(predictions)
            df.columns = output_cols  # Rename columns
        elif isinstance(predictions[0], list):
            # df = pd.DataFrame(predictions, columns=output_cols)
            # Handle the case where predictions are provided as lists
            preds_data = predictions[0]
            stds_data = predictions[1] if len(predictions) > 1 else None
            
            # Create DataFrame for predictions
            df_preds = pd.DataFrame(preds_data, columns=output_cols)
            
            if stds_data is not None:
                # Create DataFrame for standard deviations if available
                df_stds = pd.DataFrame(stds_data, columns=output_std_cols)
                # Combine DataFrames
                df = pd.concat([df_preds, df_stds], axis=1)
            else:
                df = df_preds
        elif isinstance(predictions, list) and isinstance(predictions[0], (int, float)):
            # Handling pre_data (list of numeric values)
            df = pd.DataFrame(predictions, columns=output_cols)
            print("Data is pre_data (list of numbers)")
        
        else:
            raise ValueError("Unsupported data structure in 'predictions'.")
        return df
    
```

```python
import subprocess
import json


'''this script is written to run an automated test for All AI_ML widget widget except auto-ml
and LGBM'''

CONDA_INIT_SCRIPT = "/opt/anaconda3/etc/profile.d/conda.sh"

# Define test files and corresponding folders
all_test_file_name = ['extra_trees', 'linear_regression', 'mobo', 'rescale', "mpr",
                      "xgboost", "nnfastai", "nn_torch", "active_learning", "catboost", 
                      "rf", "svm", "gpr", "gpc", "knn"]
all_test_folder_name = ['extra_trees', 'linear_regression', 'mobo', 'rescale', 
                        'multi_polynomial_regression', "xgboost", "nnfastai", "nn_torch",
                        "active_learning", "catboost", "randomforest", "svm", "gpr", 
                        "gaussian_process_cls", "k_nearest_neighbors"]

# Initialize environment name
env_name = 'hexaind3_ml_environment'

# Store results
results = []

# Iterate through all tests
for index, test_file_name in enumerate(all_test_file_name):
    # Adjust environment based on test
    if test_file_name in ['mobo', 'rescale', 'active_learning']:
        env_name = 'hexaind3_environment'
    else:
        env_name = 'hexaind3_ml_environment'

    # Test file path
    test_file_path = f"app/tests/services/AI/{all_test_folder_name[index]}/test_{test_file_name}.py"
    try:
        # Run pytest command
        result = subprocess.run(
            f"source {CONDA_INIT_SCRIPT} && conda run -n {env_name} pytest {test_file_path}",
            shell=True,
            check=True,
            executable="/bin/bash",
            capture_output=True,
            text=True
        )
        # Store success result
        results.append({
            "test_file": test_file_name,
            "environment": env_name,
            "status": "PASS",
            "output": result.stdout.strip()
        })
    except subprocess.CalledProcessError as e:
        # Store failure or error result
        results.append({
            "test_file": test_file_name,
            "environment": env_name,
            "status": "FAIL",
            "output": e.stdout.strip(),
            "error": e.stderr.strip()
        })

# Save results to a JSON file
output_file = "test_results.json"
with open(output_file, "w") as f:
    json.dump(results, f, indent=4)

# Print summary to console
print("Test Summary:")
for result in results:
    print(f"Test File: {result['test_file']}, Environment: {result['environment']}, Status: {result['status']}")
    if result['status'] == "FAIL":
        print(f"Error: {result.get('error', 'No details')}")
    print("-" * 50)

print(f"Detailed results saved to {output_file}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_iqr(data):
    # Convert the data to a numpy array for easy calculation
    data = np.array(data)
    
    # Calculate the first and third quartiles (Q1 and Q3)
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    
    # Calculate the IQR
    IQR = Q3 - Q1
    return IQR, Q1, Q3

def plot_boxplot(data, color_by=None):
    # Calculate IQR and quartiles
    iqr, Q1, Q3 = calculate_iqr(data)
    
    # Determine colors
    if color_by == 'outliers':
        # Find outliers: data points that fall outside the range of [Q1 - 1.5*IQR, Q3 + 1.5*IQR]
        lower_bound = Q1 - 1.5 * iqr
        upper_bound = Q3 + 1.5 * iqr
        colors = ['red' if x < lower_bound or x > upper_bound else 'blue' for x in data]
    elif color_by == 'all_blue':
        colors = ['blue'] * len(data)
    else:
        colors = ['grey'] * len(data)  # Default color if no option specified
    
    # Create box plot with custom color
    fig, ax = plt.subplots()
    ax.boxplot(data, vert=False, patch_artist=True, 
               boxprops=dict(facecolor='lightblue', color='black'),
               whiskerprops=dict(color='black'),
               capprops=dict(color='black'),
               flierprops=dict(marker='o', color='red', markersize=10))  # Outliers in red
    ax.set_title("Box Plot with IQR")
    ax.set_xlabel("Values")
    
    # Scatter plot to highlight individual data points with custom colors
    ax.scatter(data, [1]*len(data), color=colors, label='Data points')
    
    plt.show()

# Example usage
data = [12, 15, 14, 10, 17, 19, 25, 23, 18, 30]
plot_boxplot(data, color_by='outliers')  # Change 'outliers' to 'all_blue' for a different color scheme
```
