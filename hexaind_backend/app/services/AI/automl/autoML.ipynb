ML flow integrated code

```python

import mlflow
from autogluon.tabular import TabularDataset, TabularPredictor
from sklearn.preprocessing import StandardScaler
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Configuring plot settings
plt.rc('xtick', labelsize=20)
plt.rc('ytick', labelsize=20)
font = {'family': 'serif', 'weight': 'normal', 'size': 22}
plt.rc('font', **font)

# MLflow setup
mlflow.set_tracking_uri("http://localhost:5000")  # Update this URI to your MLflow server's URI
mlflow.set_experiment("AutoGluon_Tabular_Predictions")

# Data loading and preprocessing
data = TabularDataset('/home/muhammad/Downloads/k_data.csv')
data = data.sample(frac=1).reset_index(drop=True)
train_data = data.sample(frac=0.8, random_state=42)
test_data = data.drop(train_data.index)
numeric_features = train_data.select_dtypes(include='number').columns.tolist()

scaler = StandardScaler()
train_data[numeric_features] = scaler.fit_transform(train_data[numeric_features])
test_data[numeric_features] = scaler.transform(test_data[numeric_features])

label = 'k33'

# Hyperparameters
hyperparameters = {
    'GBM': {},
    'CAT': {},
    'XGB': {},
    'RF': {},
    'XT': {},
    'KNN': {},
    'LR': {},
    'NN_TORCH': {},
    'FASTAI': {},
}

hyperparameter_tune_kwargs = {
    'scheduler': 'local',
    'searcher': 'auto',
    'num_trials': 1,
    'time_out': 600,
}

model_path = "autogluon_model"

with mlflow.start_run():
    mlflow.log_params(hyperparameters)
    mlflow.log_params(hyperparameter_tune_kwargs)

    predictor = TabularPredictor(label=label, path=model_path).fit(
        train_data,
        fit_weighted_ensemble=False,
        fit_full_last_level_weighted_ensemble=False,
        hyperparameters=hyperparameters,
        hyperparameter_tune_kwargs=hyperparameter_tune_kwargs
    )
    
    y_pred_test = predictor.predict(test_data.drop(columns=[label]))
    
    test_metrics = predictor.evaluate(test_data, silent=True)
    mlflow.log_metrics(test_metrics)
    predictor.save()    
    mlflow.log_artifacts(model_path, artifact_path="model")
    extra_metrics = ['mean_absolute_error', 'mean_squared_error', 'root_mean_squared_error']
    summary = predictor.leaderboard(test_data, extra_metrics=extra_metrics)
    # summary = predictor.leaderboard(test_data)
    summary_file = 'summary.csv'
    summary.to_csv(summary_file)
    mlflow.log_artifact(summary_file)
    
    additional_metrics = {
        'MAE': mean_absolute_error(test_data[label], y_pred_test),
        'MSE': mean_squared_error(test_data[label], y_pred_test),
        'R2': r2_score(test_data[label], y_pred_test)
    }
    mlflow.log_metrics(additional_metrics)
```
